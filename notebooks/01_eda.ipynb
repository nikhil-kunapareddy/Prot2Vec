# 01 — Exploratory Data Analysis

Load Pfam families and visualize amino acid composition with PCA and UMAP.

```python
import logging
logging.basicConfig(level=logging.INFO, format='%(levelname)s %(message)s')

from prot2vec.data.pfam import download_pfam_seed, parse_pfam_families
from prot2vec.data.dataset import ProteinDataset

PFAM_IDS = ["PF00069", "PF00072"]

seed_path = download_pfam_seed(version="35.0", cache_dir="../data/raw")
records = parse_pfam_families(PFAM_IDS, seed_path)
dataset = ProteinDataset.from_pfam_records(records, min_length=50)
print(dataset)
```

```python
import pandas as pd
from prot2vec.embedders.composition import CompositionEmbedder

X_comp = CompositionEmbedder().fit_transform(dataset.sequences)

AA_LIST = list("ACDEFGHIKLMNPQRSTVWY")
df = pd.DataFrame(X_comp, columns=AA_LIST)
df["family"] = dataset.labels
df.head()
```

```python
from prot2vec.reduction.reducers import PCAReducer, UMAPReducer
from prot2vec.visualization.plots import scatter_2d

X_pca = PCAReducer(n_components=2).fit_transform(X_comp)
scatter_2d(X_pca, dataset.labels, "PCA of Amino Acid Composition", dim_labels=("PC1", "PC2"))
```

```python
X_umap = UMAPReducer(n_neighbors=10, min_dist=0.1).fit_transform(X_comp)
scatter_2d(X_umap, dataset.labels, "UMAP of Amino Acid Composition", dim_labels=("UMAP-1", "UMAP-2"))
```
